## Task 1

```python

def new_number(number) -> None:
    print(f"Received a new number: {number}")


def is_positive(number) -> None:
    if number > 0:
        print(f"{number} is a positive number")
    elif number == 0:
        print("Zero")
    elif number < 0:
        print(f"{number} is a negative number") 

def print_bye(number) -> None:
    print("Bye!")



def numbers_handler(numbers, before=new_number, action=is_positive, after=print_bye) -> None:

    for i in numbers:
        before(i)
        action(i)
        after(i)
numbers_handler([1, -1, 0])

def print_hello(number):
    print("Hello!")

def some_action(number):
    print("Action!")

numbers_handler([10], action=some_action, before=print_hello)
```

## Task 2

```python
def get_average_mark(student_name, *marks):
    
    if not student_name or student_name.strip() == "":
        return "Invalid student name"
    
    if not marks:
        return "Student has no marks"
    
    for i in marks:
        if i < 0:
            return "Invalid marks"
    
    average = round(sum(marks) / len(marks))
    return f"{student_name} got {average}"
```

```python
print(get_average_mark("Danylo", 11))
print(get_average_mark("Oleksii", 12, 10, 11, 11, 11))
print(get_average_mark("Ivan", 10, 11, 12, 9, 12, 12))
```

## Task 3

```python
def message_people(people):
    if not people:
        raise ValueError("Empty list people")

    def print_message(message):
        names_join = ", ".join(people)
        
        if message == 'hello':
            print(f"Hello, {names_join}, nice to see you!")
        elif message == 'meeting':
            print(f"{names_join}, we have a meeting in an hour, don't be late!")
        elif message == 'bye':
            print(f"Bye {names_join}, see you tomorrow!")
        else:
            print("Unknown message type")
    
    return print_message
    
mes_people = message_people(["Alex", "Robert", "Tom"])
mes_people('hello')
mes_people('meeting')
mes_people('bye')
```

## Task 4

```python
def fibonacci_generator(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b
```

```python
fib_generator = fibonacci_generator(10)
for fib_number in fib_generator:
    print(fib_number)
```

## Task 5

```python
class Person:
    people = {}

    def __init__(self, name, age):
        self.name = name
        self.age = age
        Person.people[name] = self

def create_person_list(people):
    person_list = []
    
    for p in people:
        name = p['name']
        age = p['age']
        person = Person(name, age)
        person_list.append(person)
    
    for p in people:
        current = Person.people[p['name']]
        if 'wife' in p and p['wife'] is not None:
            current.wife = Person.people[p['wife']]
        if 'husband' in p and p['husband'] is not None:
            current.husband = Person.people[p['husband']]
    
    return person_list




    
```

```python
people = [
    {'name': 'Ross', 'age': 30, 'wife': 'Rachel'},
    {'name': 'Joey', 'age': 29, 'wife': None},
    {'name': 'Rachel', 'age': 28, 'husband': 'Ross'}
]

person_list = create_person_list(people)

print(isinstance(person_list[0], Person)) # True
print(person_list[0].name == 'Ross')
print(person_list[0].wife is person_list[2]) # True
print(person_list[0].wife.name == 'Rachel')
print()
print(person_list[1].name == 'Joey')
print()
print(isinstance(person_list[2], Person)) # True
print(person_list[2].name == 'Rachel')
print(person_list[2].husband is person_list[0]) # True
print()

print(person_list[2].husband.name == 'Ross')
print(person_list[2].husband.wife is person_list[2])  # True
print(Person.people)
```
